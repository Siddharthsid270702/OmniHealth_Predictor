```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.utils.class_weight import compute_class_weight
import numpy as np

# Load your dataset
df = pd.read_csv("C:\\Users\\siddharth\\Desktop\\sem4 capstone code\\test\\uniqueDatasets\\training.csv")

# Split data into features and target
X = df.iloc[:, 1:]  # Features (Symptom columns)
y = df.iloc[:, 0]   # Target (Disease column)

# Handle categorical features if any, encode them
X = X.apply(LabelEncoder().fit_transform)

# Encode the target labels
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)

# Split dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.2, random_state=42)

# Feature scaling
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

# Compute class weights
class_weights = compute_class_weight('balanced', classes=np.unique(y_encoded), y=y_encoded)
class_weight_dict = dict(zip(np.unique(y_encoded), class_weights))

# Initialize and train Random Forest Classifier
rf = RandomForestClassifier(n_estimators=100, random_state=42, class_weight=class_weight_dict)
rf.fit(X_train, y_train)

# Make predictions
y_pred_rf = rf.predict(X_test)

# Evaluate model
accuracy_rf = accuracy_score(y_test, y_pred_rf)
print(f"Random Forest Accuracy: {accuracy_rf:.4f}")
print(f"Random Forest Classification Report:\n", classification_report(y_test, y_pred_rf))
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier, AdaBoostClassifier, BaggingClassifier, StackingClassifier, VotingClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.utils.class_weight import compute_class_weight
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.linear_model import LogisticRegression
import numpy as np
import joblib
import pickle

# Load your dataset (replace with your actual file path)
df = pd.read_csv("C:\\Users\\siddharth\\Desktop\\sem4 capstone code\\test\\uniqueDatasets\\training.csv")

# Split data into features and target
X = df.iloc[:, 1:]  # Features (Symptom columns)
y = df.iloc[:, 0]   # Target (Disease column)

# Encode categorical features if any
X = X.apply(LabelEncoder().fit_transform)

# Encode the target labels
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)

# Feature scaling
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Compute class weights to handle class imbalance
class_weights = compute_class_weight('balanced', classes=np.unique(y_encoded), y=y_encoded)
class_weight_dict = dict(zip(np.unique(y_encoded), class_weights))

# Initialize classifiers
models = {
    "Random Forest": RandomForestClassifier(n_estimators=100, random_state=42, class_weight=class_weight_dict),
    "Naive Bayes": GaussianNB(),
    "Gradient Boosting": GradientBoostingClassifier(n_estimators=100, random_state=42),
    "AdaBoost": AdaBoostClassifier(n_estimators=100, random_state=42),
    "Bagging": BaggingClassifier(estimator=RandomForestClassifier(), n_estimators=10, random_state=42),
    "Stacking": StackingClassifier(
        estimators=[
            ('svm', SVC(kernel='linear', probability=True, random_state=42)),
            ('rf', RandomForestClassifier(n_estimators=100, random_state=42)),
            ('dt', DecisionTreeClassifier(random_state=42))
        ],
        final_estimator=LogisticRegression()
    ),
    "Voting": VotingClassifier(
        estimators=[
            ('rf', RandomForestClassifier(n_estimators=100, random_state=42)),
            ('adaboost', AdaBoostClassifier(n_estimators=100, random_state=42)),
            ('logreg', LogisticRegression())
        ],
        voting='hard'
    )
}

# Train and evaluate models for different train-test splits
splits = [0.9, 0.8, 0.7, 0.6, 0.5]
results = {}
best_model_instance = None
best_model_name = None
best_accuracy = 0

for split in splits:
    print(f"\nEvaluating for {int(split * 100)}% train / {int((1 - split) * 100)}% test split...")
    
    # Split dataset
    X_train, X_test, y_train, y_test = train_test_split(X_scaled, y_encoded, test_size=(1 - split), random_state=42)

    # Store accuracy for this split
    split_results = {}

    for model_name, model in models.items():
        print(f"\nTraining and evaluating {model_name}...")
        model.fit(X_train, y_train)
        
        # Make predictions
        y_pred = model.predict(X_test)
        
        # Evaluate model
        accuracy = accuracy_score(y_test, y_pred)
        split_results[model_name] = accuracy
        print(f"{model_name} Accuracy: {accuracy:.4f}")
        print(f"{model_name} Classification Report:\n", classification_report(y_test, y_pred))
        
        # Check if this model is the best so far
        if accuracy > best_accuracy:
            best_accuracy = accuracy
            best_model_instance = model
            best_model_name = model_name
    
    # Store split results
    results[split] = split_results

print(f"\nBest performing model: {best_model_name} - Accuracy: {best_accuracy:.4f}")

# Save the best performing model
if best_model_instance:
    joblib.dump(best_model_instance, 'best_model.pkl')
    print("The best performing model has been saved as 'best_model.pkl'.")

# Save the label encoder for later use in the web application
if label_encoder:
    with open('label_encoder.pkl', 'wb') as file:
        pickle.dump(label_encoder, file)
    print("The label encoder has been saved as 'label_encoder.pkl'.")
```

```python
import matplotlib.pyplot as plt
import numpy as np
import joblib
from sklearn.metrics import roc_curve, auc
from sklearn.preprocessing import label_binarize
from sklearn.model_selection import train_test_split

# Load pre-trained model
best_model = joblib.load("best_model.pkl")

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y_encoded, test_size=0.3, random_state=42)

# Get unique class count
n_classes = len(np.unique(y_encoded))

# One-hot encode labels only if multi-class
if n_classes > 2:
    y_train_bin = label_binarize(y_train, classes=np.arange(n_classes))
    y_test_bin = label_binarize(y_test, classes=np.arange(n_classes))
else:
    y_train_bin = y_train.reshape(-1, 1)  # Convert to 2D array for binary classification
    y_test_bin = y_test.reshape(-1, 1)

# Get probability predictions
y_train_score = best_model.predict_proba(X_train)
y_test_score = best_model.predict_proba(X_test)

# Handle case where only one class exists
if y_train_bin.shape[1] == 1:  
    plt.figure(figsize=(6, 6))
    fpr_train, tpr_train, _ = roc_curve(y_train_bin, y_train_score[:, 1])
    fpr_test, tpr_test, _ = roc_curve(y_test_bin, y_test_score[:, 1])
    
    roc_auc_train = auc(fpr_train, tpr_train)
    roc_auc_test = auc(fpr_test, tpr_test)

    plt.plot(fpr_train, tpr_train, linestyle="dashed", label=f'Train (AUC = {roc_auc_train:.3f})')
    plt.plot(fpr_test, tpr_test, label=f'Test (AUC = {roc_auc_test:.3f})')

else:
    # Multi-class case: Plot ROC curve for each class
    plt.figure(figsize=(6, 6))
    for i in range(y_train_bin.shape[1]):
        fpr_train, tpr_train, _ = roc_curve(y_train_bin[:, i], y_train_score[:, i])
        fpr_test, tpr_test, _ = roc_curve(y_test_bin[:, i], y_test_score[:, i])
        
        roc_auc_train = auc(fpr_train, tpr_train)
        roc_auc_test = auc(fpr_test, tpr_test)

        plt.plot(fpr_train, tpr_train, linestyle="dashed", label=f'Train Class {i} (AUC = {roc_auc_train:.3f})')
        plt.plot(fpr_test, tpr_test, label=f'Test Class {i} (AUC = {roc_auc_test:.3f})')

# Random baseline
plt.plot([0, 1], [0, 1], 'k--')

plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve: Train vs Test')
plt.legend()
plt.show()
```
